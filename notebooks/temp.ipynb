```python
particles = jax.random.normal(jax.random.PRNGKey(0), (3, 2))

y = jnp.array([[1.0], [2.0], [3.0]])

sigma = 0.5

parameters = ProParameters(
    y=y,
    step_size=0.1,
    sigma=sigma,
    alpha=1.0,
    tolerance=1e-300,
)

normal_logpdf(y, particles, sigma)
```

```python
marginal = pro_replica_fn(particles, parameters)
jnp.exp(marginal).sum(axis=1)
```

```python
dens = jnp.exp(normal_logpdf(y, particles, sigma))
dens = dens / jnp.sum(dens, axis=1, keepdims=True)
dens.sum(axis=1)
```
